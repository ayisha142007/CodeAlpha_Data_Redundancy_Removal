```python
import pandas as pd


data = {
    "Customer_ID": [101, 102, 103, 104, 105, 102, 106],
    "Name": ["Aisha", "Rahul", "Priya", "Kumar", "Sara", "Rahul", "Arun"],
    "Email": [
        "aisha@gmail.com",
        "rahul@gmail.com",
        "priya@gmail.com",
        "kumar@gmail.com",
        "sara@gmail.com",
        "rahul@gmail.com",
        "arun@gmail.com"
    ]
}

df = pd.DataFrame(data)


df.to_csv("customer_data.csv", index=False)


df
```

```python


duplicates = df[df.duplicated(keep=False)]

print("Duplicate Records:")
duplicates
```

```python


df_cleaned = df.drop_duplicates()

print("Original number of records:", len(df))
print("Records after removing duplicates:", len(df_cleaned))

df_cleaned
```

```python


new_customer = {
    "Customer_ID": 107,
    "Name": "Meena",
    "Email": "meena@gmail.com"
}


if (new_customer["Customer_ID"] in df_cleaned["Customer_ID"].values or
    new_customer["Email"] in df_cleaned["Email"].values):

    print("Duplicate data found! Customer cannot be added.")

else:
    df_cleaned = pd.concat(
        [df_cleaned, pd.DataFrame([new_customer])],
        ignore_index=True
    )
    print("New customer is unique and has been added.")

df_cleaned
```

```python


df_cleaned.to_csv("final_customer_data.csv", index=False)

print("Final customer data saved successfully!")
```

```python


print("Total records:", len(df_cleaned))
print("Duplicate records:", df_cleaned.duplicated().sum())
print("\nFinal Dataset:")
display(df_cleaned)
```
